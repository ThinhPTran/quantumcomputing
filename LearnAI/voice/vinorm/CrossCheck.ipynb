```python
def read_first_column(filepath):
    """
    Reads the first column from a file and returns it as a list.

    Args:
        filepath: The path to the input file.

    Returns:
        A list of strings, where each string is an item from the first column.
    """
    first_column_list = []
    try:
        with open(filepath, 'r', encoding='utf-8') as file:
            for line in file:
                # Split the line by whitespace and take the first element
                parts = line.split()
                if parts:  # Ensure the line is not empty
                    first_column_list.append(parts[0])
    except FileNotFoundError:
        print(f"Error: The file '{filepath}' was not found.")
        return None
    return first_column_list

# Specify the path to your data file
file_to_read = 'mytokens.txt'

# Call the function and get the list
mytokens = read_first_column(file_to_read)

# Print the result
if mytokens:
    print("Successfully read the first column into a list:")
    print(len(mytokens))
    print(mytokens)
```

```python
# --- Main script execution ---

# Assume you already have these two lists in your program
list1 = syms
list2 = mytokens

print(f"List 1: {list1}")
print(f"List 2: {list2}\n")

# Get the unique set out of each list
set1 = set(list1)
set2 = set(list2)

print(f"Unique elements from list 1 (Set 1): {set1}")
print(f"Unique elements from list 2 (Set 2): {set2}\n")

# Check if the two sets are equal
if set1 == set2:
    print("Result: The two sets are equal.")
else:
    print("Result: The two sets are not equal.")

    # Find elements that are unique to set1
    only_in_set1 = set1.difference(set2)
    
    # Find elements that are unique to set2
    only_in_set2 = set2.difference(set1)

    if only_in_set1:
        print("\nElements found only in the first set:")
        # sorted() is used here to make the output order consistent
        for element in sorted(list(only_in_set1)):
            print(f"- '{element}'")

    if only_in_set2:
        print("\nElements found only in the second set:")
        # sorted() is used here to make the output order consistent
        for element in sorted(list(only_in_set2)):
            print(f"- '{element}'")
```
