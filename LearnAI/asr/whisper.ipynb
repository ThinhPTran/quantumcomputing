```python
import whisper
import time
from pydub import AudioSegment
import math
import os

# --- Setup ---
# NOTE: "turbo" is not a standard model name. Using "base" instead.
# Other options: "tiny", "small", "medium", "large"
MODEL_NAME = "turbo"
AUDIO_FILE = "./sample/1. DANG VU KHOA 2813917876.wav"
SEGMENT_DURATION_S = 15  # Process in 25-second chunks

# --- Main Script ---

# 1. Load Whisper Model
print(f"Loading Whisper model ('{MODEL_NAME}')...")
try:
    model = whisper.load_model(MODEL_NAME)
    print("Model loaded successfully.")
except Exception as e:
    print(f"Error loading Whisper model: {e}")
    exit()

# 2. Load Audio File with pydub
print(f"Loading audio file: {AUDIO_FILE}")
try:
    audio = AudioSegment.from_wav(AUDIO_FILE)
except FileNotFoundError:
    print(f"Error: The audio file was not found at '{AUDIO_FILE}'")
    exit()

# 3. Process Audio in Segments
print("-" * 50)
print(f"Starting transcription in {SEGMENT_DURATION_S}s segments...")
start_time = time.time()

# Convert segment duration to milliseconds for pydub
segment_length_ms = SEGMENT_DURATION_S * 1000
full_transcription = []
temp_file_name = "temp_segment.wav"

# Calculate the total number of segments
num_segments = math.ceil(len(audio) / segment_length_ms)

for i in range(num_segments):
    start_ms = i * segment_length_ms
    end_ms = start_ms + segment_length_ms
    
    # Extract the segment
    segment = audio[start_ms:end_ms]
    
    # Export the segment to a temporary file
    segment.export(temp_file_name, format="wav")
    
    # Print progress
    start_s = start_ms // 1000
    print(f"\n[+] Processing segment {i+1}/{num_segments} ({start_s}s to {end_ms // 1000}s)...")
    
    # Transcribe the segment file
    result = model.transcribe(temp_file_name, word_timestamps=False) # Word timestamps not needed for segment text
    segment_text = result["text"].strip()
    
    print(f"  -> Transcription: {segment_text}")
    
    # Append the result to our full list
    full_transcription.append(segment_text)

# Clean up the temporary file
if os.path.exists(temp_file_name):
    os.remove(temp_file_name)

end_time = time.time()
duration = end_time - start_time

# 4. Print Final Results
print("-" * 50)
print("✅ Transcription Complete!")
print("\nFull Combined Transcription:")
print(" ".join(full_transcription))

print("-" * 50)
print(f"Total transcription process took {duration:.2f} seconds.")
```

```python
import torch
from transformers import AutoModelForSpeechSeq2Seq, AutoProcessor, pipeline
import time

device = "mps"
torch_dtype = torch.float16 if torch.cuda.is_available() else torch.float32

model_id = "openai/whisper-large-v3-turbo"

model = AutoModelForSpeechSeq2Seq.from_pretrained(
    model_id, torch_dtype=torch_dtype, low_cpu_mem_usage=True, use_safetensors=True
)
model.to(device)

processor = AutoProcessor.from_pretrained(model_id)

pipe = pipeline(
    "automatic-speech-recognition",
    model=model,
    tokenizer=processor.tokenizer,
    feature_extractor=processor.feature_extractor,
    max_new_tokens=128,
    chunk_length_s=25,
    batch_size=16,
    torch_dtype=torch_dtype,
    device=device,
)

start_time = time.time()
result = pipe("./sample/1. DANG VU KHOA 2813917876.wav")


print(result["text"])
end_time = time.time()
duration = end_time - start_time

print("-" * 50)
print(f"Total transcription process took {duration:.2f} seconds.")
```
