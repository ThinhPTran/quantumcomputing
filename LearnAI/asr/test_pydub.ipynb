```python
import numpy as np
from pydub import AudioSegment
from IPython.display import Audio

# Load your audio file
song = AudioSegment.from_file("sample/emlinh_5s.mp3")

# Slice the audio (e.g., from 1.5s to 3.5s)
start_time_ms = 4720
end_time_ms = 5000
extracted_clip = song[start_time_ms:end_time_ms]

# Get the raw audio data as a NumPy array
# Pydub samples are integers, so we convert them to floating point for the player
samples = np.array(extracted_clip.get_array_of_samples()).astype(np.float32)

# Normalize the audio to be between -1 and 1
# This is required by the IPython Audio player for float data
samples /= np.iinfo(extracted_clip.sample_width * 8).max

# Play the audio directly in the notebook
# You MUST provide the sample rate (extracted_clip.frame_rate)
Audio(samples, rate=extracted_clip.frame_rate)
```
